# 3 - Baseline CNN

A simple from-scratch CNN to establish a performance baseline.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.config import Config
from src.utils import set_global_seed
from src.data_loader import create_all_pipelines, compute_class_weights
from src.models.baseline_cnn import build_baseline_cnn
from src.training import train_model
from src.evaluation import evaluate_model

cfg = Config()
set_global_seed(cfg.seed)

train_ds, val_ds, test_ds = create_all_pipelines(cfg)
class_weights = compute_class_weights(cfg)
```

```python
history = train_model(
    model,
    train_ds,
    val_ds,
    cfg,
    experiment_name='01_baseline_cnn',
    class_weight=class_weights,
)
```
