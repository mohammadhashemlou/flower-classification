# 6 - Results Comparison

Compare all models side-by-side.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
from src.config import Config

cfg = Config()

experiments = {
    'Baseline CNN': cfg.experiment_dir('01_baseline_cnn') / 'history.csv',
    'Residual CNN': cfg.experiment_dir('02_residual_cnn') / 'history.csv',
    'Transfer (Phase 1)': cfg.experiment_dir('03_efficientnet_frozen') / 'history.csv',
    'Transfer (Phase 2)': cfg.experiment_dir('04_efficientnet_finetuned') / 'history.csv',
}

results = []
for name, path in experiments.items():
    if path.exists():
        df = pd.read_csv(path)
        results.append({
            'Model': name,
            'Best Val Acc': df['val_accuracy'].max(),
            'Best Val Loss': df['val_loss'].min(),
            'Epochs': len(df),
        })

results_df = pd.DataFrame(results)
display(results_df)

# Plot comparison
fig, ax = plt.subplots(figsize=(10, 5))
for name, path in experiments.items():
    if path.exists():
        df = pd.read_csv(path)
        ax.plot(df['val_accuracy'], label=name, linewidth=1.5)

ax.set_title('Validation Accuracy - All Models')
ax.set_xlabel('Epoch')
ax.set_ylabel('Val Accuracy')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
